#Praktikum: Toko Dorayaki Doraemon

Program ini dibuat untuk membantu Doraemon dkk dalam membuat nama toko dorayaki. Program menerima input berupa nama makanan, nama karakter, dan kata sifat, lalu menggabungkannya menjadi nama toko.

Selain itu, program digunakan untuk menghitung total modal dari Doraemon, Nobigon, dan Shizuka, menghitung pendapatan dari penjualan dorayaki pada bulan pertama dan menentukan keuntungan yang diperoleh toko.


```python
# 1
# Meminta nama makanan dari pengguna
nama_makanan = input("Masukkan nama makanan: ")
# Meminta nama karakter dari pengguna
nama_karakter = input("Masukkan nama karakter: ")
# Meminta kata sifat dari pengguna
kata_sifat = input("Masukkan kata sifat: ")

#2
# Menggabungkan ketiga input menjadi nama toko
nama_toko = nama_makanan + " " + nama_karakter + " " + kata_sifat
# Menampilkan nama toko
print("Nama resmi toko:", nama_toko)

#3
# Menyimpan modal yang diberikan Doraemon
modal_doraemon = 555000
# Menyimpan modal yang diberikan Nobita
modal_nobita = 333000
# Menyimpan modal yang diberikan Shizuka
modal_shizuka = 222000

#4
# Menghitung keseluruhan modal
total_modal = modal_doraemon + modal_nobita + modal_shizuka
# Menampilkan total modal
print("Total modal: Rp", total_modal)

#5
# Menyimpan jumlah dorayaki yang terjual
jumlah_dorayaki = 350
# Menyimpan harga setiap dorayaki
harga_dorayaki = 5000
# Menghitung pendapatan dari penjualan dorayaki
total_pendapatan = jumlah_dorayaki * harga_dorayaki
# Menampilkan total pendapatan
print("Total pendapatan: Rp", total_pendapatan)

#6
# Mengurangi pendapatan dengan modal untuk mendapatkan keuntungan
keuntungan = total_pendapatan - total_modal
# Menampilkan keuntungan bulan pertama
print("Keuntungan bulan pertama: Rp", keuntungan)

#7
# error Nobita:
# Nobigon menulis: print("Keuntungan total adalah" keuntungan)
# Error karena kesalahan tanda koma (,) atau tanda + hilang diantara string dan variabel.
# yang benar adalah: print("Keuntungan total adalah", keuntungan) atau print("Keuntungan total adalah " + str(keuntungan))
```
