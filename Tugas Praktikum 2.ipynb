```python
# 1. Menyimpan data diri Giant
nama = "Giant"
usia = 20
berat = "97.5"
tinggi = 1.75
status_olahraga = False

# 2. Mengubah berat menjadi float dan membuat kalimat pengantar
berat = float(berat)
print(f"Halo, saya {nama}. Usia saya {usia} tahun dan berat saya {berat} kg")

# 3. Menghitung BMI Giant
bmi = berat / (tinggi ** 2)
print("BMI asli:", bmi)

bmi_bulat = round(bmi, 2)
print("BMI setelah dibulatkan:", bmi_bulat)

# 4. Menghitung kebutuhan air minum
target_air = 2000
kapasitas_botol = 350

# True division
jumlah_botol_asli = target_air / kapasitas_botol
print("Jumlah botol asli:", jumlah_botol_asli)

# Floor division
botol_penuh = target_air // kapasitas_botol
print("Botol penuh:", botol_penuh)

# Modulo
sisa_air = target_air % kapasitas_botol
print("Sisa air:", sisa_air, "ml")


# 5. Menghitung kalori dengan floor() dan ceil()
import math

kalori_harian = 2100.5
# Pembulatan ke bawah
kalori_bawah = math.floor(kalori_harian)
# Pembulatan ke atas
kalori_atas = math.ceil(kalori_harian)

print("Kalori dibulatkan ke bawah:", kalori_bawah)
print("Kalori dibulatkan ke atas:", kalori_atas)
```
